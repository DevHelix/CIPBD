```python
import deepdoctection as dd
import csv
import re
from IPython.display import display, HTML
from matplotlib import pyplot as plt
```

```python
analyzer = dd.get_dd_analyzer()  # instantiate the built-in analyzer similar to the Hugging Face space demo
```

```python
df = analyzer.analyze(path = r"C:\Users\vince\Documents\GitHub\CIPBD\Dallas\PDF\2024.pdf")  # setting up pipeline
df.reset_state()                 # Trigger some initialization
```

```python
doc = iter(df)

pages = []
for i in range(len(pdf.pages)):  # total page count from pdfplumber
    try:
        page = next(doc)
        pages.append(page)
    except ValidationError:
        print(f"Skipping page {i+1} (validation error)")
        continue
    except StopIteration:
        break
```

```python

ids = []
count = -1

def page_text(table_csv):
    """Flatten a table's csv (list of lists) into one searchable string."""
    return ' '.join(cell for row in table_csv for cell in row if cell)

for page in pages:
    count = count + 1
    text = page_text(page.tables[0].csv)

    has_keyword1 = "Council District" in text
    has_keyword2 = "Service" in text

    if has_keyword1 and has_keyword2:
        ids.append(count)

print(ids)
```

```python
with open("output.csv", "a", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(pages[0].tables[0].csv)
    for id in ids:
        page = pages[0]
        writer.writerows(page.tables[0].csv[1:])
```
